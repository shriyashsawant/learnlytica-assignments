# Capstone Mini-Project: Predict Customer Churn (End-to-End Tabular ML)

**Business question.** Which customers are about to leave, and what should we do about it?

**Problem framing.** Given a customer's account and plan features *as of today*, predict whether they will churn (`churn = 1`), so the retention team can target save-offers. This is binary classification on imbalanced classes (about 27% positive).

**Metric choice.** A churner we miss costs far more than a happy customer we call, so we protect **recall on the churn class**. We use **ROC-AUC** (threshold independent) to select the model.

**Plan.** 1) explore, 2) split and hunt for leakage, 3) leakage-free Pipeline, 4) compare four model families, 5) tune the winner, 6) evaluate once on the test set and choose a threshold, 7) interpret, 8) recommend.

## Prerequisites: synthesise the dataset (in memory, no download)

```python
import warnings; warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification

RNG = 42
rng = np.random.default_rng(RNG)
N = 4000

# Latent signal: 5 informative directions separating churners from stayers.
X_num, y = make_classification(
    n_samples=N, n_features=6, n_informative=5, n_redundant=0,
    n_clusters_per_class=1, weights=[0.76, 0.24], flip_y=0.13,
    class_sep=0.85, random_state=RNG)
Xs = (X_num - X_num.mean(0)) / X_num.std(0)   # standardise before rescaling so signal survives

df = pd.DataFrame()
# --- numeric account features (rescaled to human-readable ranges) ---
df["tenure_months"]    = np.clip(Xs[:, 4] * 12 + 34, 1, 72).round().astype(int)
df["monthly_charges"]  = np.clip(Xs[:, 0] * 13 + 72, 18, 130).round(2)
df["support_tickets"]  = np.clip(Xs[:, 2] * 1.7 + 3, 0, 25).round().astype(int)
df["avg_latency_ms"]   = np.clip(Xs[:, 3] * 35 + 120, 20, 400).round().astype(int)
df["addons_count"]     = np.clip(Xs[:, 1] * 1.3 + 4, 0, 8).round().astype(int)
df["days_since_login"] = np.clip(-Xs[:, 4] * 8 + 16 + rng.normal(0, 4, N), 0, 90).round().astype(int)
df["total_charges"]    = (df["monthly_charges"] * df["tenure_months"]).round(2)

# --- categorical plan features, correlated with churn like the real world ---
def by_class(p_churn, p_stay, choices):
    out = np.empty(N, dtype=object); m = (y == 1)
    out[m]  = rng.choice(choices, m.sum(),  p=p_churn)
    out[~m] = rng.choice(choices, (~m).sum(), p=p_stay)
    return out

df["contract_type"]    = by_class([0.75,0.18,0.07],[0.34,0.35,0.31],
                                  ["Month-to-month","One year","Two year"])
df["payment_method"]   = by_class([0.55,0.20,0.15,0.10],[0.25,0.30,0.25,0.20],
                                  ["Electronic check","Mailed check","Bank transfer","Credit card"])
df["internet_service"] = by_class([0.65,0.28,0.07],[0.40,0.35,0.25],
                                  ["Fiber optic","DSL","None"])
df["has_dependents"]   = by_class([0.82,0.18],[0.60,0.40], ["No","Yes"])

# --- the TRAP: a post-outcome flag that leaks the label (removed in Step 2) ---
df["exit_survey_sent"] = np.where(y == 1, rng.binomial(1,0.97,N), rng.binomial(1,0.02,N))

df["churn"] = y
print("dataset shape:", df.shape)
print("churn rate:   ", round(df["churn"].mean(), 3))
df.head()
```

## Step 1: Frame the problem and explore (EDA)
Before modelling, look at the class balance, how the numeric drivers differ between churners and stayers, and how the churn rate varies with contract type.

```python
# Class balance
print(df["churn"].value_counts(normalize=True).round(3), "\n")

# How do the numeric drivers differ between churners and stayers?
num_cols = ["tenure_months","monthly_charges","support_tickets",
            "avg_latency_ms","addons_count","days_since_login","total_charges"]
print(df.groupby("churn")[num_cols].mean().round(1).T, "\n")

# Churn rate by contract type - a categorical driver
print(df.groupby("contract_type")["churn"].mean().round(3).sort_values(ascending=False))

# Visualise tenure vs churn
fig, ax = plt.subplots(1, 2, figsize=(11,4))
for k, lbl in [(0,"stay"),(1,"churn")]:
    ax[0].hist(df.loc[df.churn==k,"tenure_months"], bins=25, alpha=0.6, label=lbl)
ax[0].set_title("Tenure by outcome"); ax[0].set_xlabel("tenure_months"); ax[0].legend()
df.groupby("contract_type")["churn"].mean().sort_values().plot.barh(ax=ax[1])
ax[1].set_title("Churn rate by contract"); ax[1].set_xlabel("churn rate")
plt.tight_layout(); plt.show()
```

## Step 2: Split train / validation / test and hunt for leakage
A test set we will not look at until Step 6, plus a validation slice, leaving the rest for training and cross-validation. Any feature whose value is only known *after* the outcome must be dropped: `exit_survey_sent` is sent because a customer already churned.

```python
from sklearn.model_selection import train_test_split

# Leakage check #1 - a feature that "knows" the answer
leak_scores = df.corr(numeric_only=True)["churn"].drop("churn").abs().sort_values(ascending=False)
print("abs. correlation with churn:\n", leak_scores.round(3), "\n")
suspects = leak_scores[leak_scores > 0.9].index.tolist()
print("LEAK suspects (corr > 0.9):", suspects)

df_clean = df.drop(columns=suspects)          # drop exit_survey_sent
X = df_clean.drop(columns=["churn"])
y = df_clean["churn"]

# 70% train / 15% validation / 15% test, stratified to keep the churn rate stable
X_train, X_tmp, y_train, y_tmp = train_test_split(
    X, y, test_size=0.30, stratify=y, random_state=RNG)
X_val, X_test, y_val, y_test = train_test_split(
    X_tmp, y_tmp, test_size=0.50, stratify=y_tmp, random_state=RNG)

print("\nshapes:", X_train.shape, X_val.shape, X_test.shape)
print("churn rate train/val/test:",
      round(y_train.mean(),3), round(y_val.mean(),3), round(y_test.mean(),3))
```

**Leakage check #2, by reasoning:** `total_charges = monthly_charges x tenure_months` is a legitimate cumulative-spend feature that is known today, so it stays. Leakage is about *timing* (is the value known at prediction time?), not about derived columns.

## Step 3: Feature engineering + scaling inside a Pipeline
Numbers are scaled and categories are one-hot encoded inside a `ColumnTransformer`, so the transformations are fit on training data only and re-applied identically to validation and test data. That is the structural cure for leakage.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder

num_features = X_train.select_dtypes("number").columns.tolist()
cat_features = X_train.select_dtypes(["object", "string"]).columns.tolist()
print("numeric    :", num_features)
print("categorical:", cat_features)

pre = ColumnTransformer([
    ("num", StandardScaler(), num_features),
    ("cat", OneHotEncoder(handle_unknown="ignore"), cat_features),
])

# Sanity-check the transform width (fit on TRAIN ONLY)
Z = pre.fit_transform(X_train)
print("\ntransformed matrix:", Z.shape)
print("feature count after one-hot:", len(pre.get_feature_names_out()))
```

## Step 4: Train and compare four model families
Four classifiers behind the same `pre` block, scored with 5-fold cross-validation on the **training set** using ROC-AUC. `class_weight="balanced"` makes the linear, margin and tree models take the minority churn class seriously.

```python
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC

models = {
    "LogReg":       LogisticRegression(max_iter=2000, class_weight="balanced"),
    "RandomForest": RandomForestClassifier(n_estimators=300, random_state=RNG,
                                           class_weight="balanced"),
    "GradBoost":    GradientBoostingClassifier(random_state=RNG),
    "SVM_RBF":      SVC(kernel="rbf", probability=True, class_weight="balanced",
                        random_state=RNG),
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RNG)
results = {}
for name, clf in models.items():
    pipe = Pipeline([("pre", pre), ("clf", clf)])
    scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring="roc_auc")
    results[name] = scores
    print(f"{name:13s} ROC-AUC = {scores.mean():.3f} +/- {scores.std():.3f}")
```

## Step 5: Tune the winner with GridSearchCV
A small, sensible grid for GradientBoosting, with the same cross-validation and scoring, so tuning never sees the validation or test sets. Then one confirmatory look at the held-out validation set.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import roc_auc_score

gb_pipe = Pipeline([("pre", pre),
                    ("clf", GradientBoostingClassifier(random_state=RNG))])

param_grid = {
    "clf__n_estimators": [150, 300],
    "clf__learning_rate": [0.05, 0.1],
    "clf__max_depth": [2, 3],
}

grid = GridSearchCV(gb_pipe, param_grid, cv=cv, scoring="roc_auc", n_jobs=-1)
grid.fit(X_train, y_train)

print("best params:", grid.best_params_)
print("best CV ROC-AUC:", round(grid.best_score_, 3))

best_model = grid.best_estimator_
val_auc = roc_auc_score(y_val, best_model.predict_proba(X_val)[:, 1])
print("validation ROC-AUC:", round(val_auc, 3))
```

## Step 6: Evaluate with the right metrics and choose a threshold
The test set is touched **exactly once** here. The default 0.5 threshold favours precision; a retention team usually wants more recall even at some precision cost, so we sweep the threshold.

```python
from sklearn.metrics import (classification_report, confusion_matrix,
                             ConfusionMatrixDisplay, roc_auc_score,
                             precision_score, recall_score, f1_score)

proba = best_model.predict_proba(X_test)[:, 1]
pred_050 = (proba >= 0.50).astype(int)

print("TEST ROC-AUC:", round(roc_auc_score(y_test, proba), 3), "\n")
print(classification_report(y_test, pred_050, target_names=["stay","churn"], digits=3))

ConfusionMatrixDisplay(confusion_matrix(y_test, pred_050),
                       display_labels=["stay","churn"]).plot(values_format="d")
plt.title("Confusion matrix @ threshold 0.50"); plt.show()

# Business trade-off: sweep the decision threshold
print(f"{'thr':>5} {'precision':>10} {'recall':>8} {'f1':>6}")
for t in [0.50, 0.40, 0.35, 0.30, 0.25]:
    p = (proba >= t).astype(int)
    print(f"{t:5.2f} {precision_score(y_test,p):10.3f} "
          f"{recall_score(y_test,p):8.3f} {f1_score(y_test,p):6.3f}")
```

## Step 7: Interpret the model
Permutation importance is model agnostic (shuffle one feature, measure the ROC-AUC drop). The logistic-regression coefficients give signed, human-readable directions.

```python
from sklearn.inspection import permutation_importance

# 1) Permutation importance of the tuned GradientBoosting model (on validation)
perm = permutation_importance(best_model, X_val, y_val,
                              n_repeats=8, random_state=RNG, scoring="roc_auc")
imp = pd.Series(perm.importances_mean, index=X.columns).sort_values(ascending=False)
print("Permutation importance (ROC-AUC drop when shuffled):")
print(imp.round(4), "\n")

# 2) Signed drivers from an interpretable LogReg baseline
logreg = Pipeline([("pre", pre),
                   ("clf", LogisticRegression(max_iter=2000,
                                              class_weight="balanced"))]).fit(X_train, y_train)
coef = pd.Series(logreg.named_steps["clf"].coef_[0],
                 index=logreg.named_steps["pre"].get_feature_names_out())
print("Most churn-DRIVING (positive coef):\n", coef.sort_values().tail(4).round(2), "\n")
print("Most PROTECTIVE (negative coef):\n", coef.sort_values().head(4).round(2))

imp.sort_values().plot.barh(figsize=(7,4), title="Permutation importance"); plt.tight_layout(); plt.show()
```

## Step 8: The recommendation (the real deliverable)
Run the scorecard cell, then read the written recommendation and reflection below it.

```python
summary = {
    "primary_metric": "recall on churn (ROC-AUC for selection)",
    "chosen_model": "GradientBoosting (tuned)",
    "test_roc_auc": round(roc_auc_score(y_test, proba), 3),
    "operating_threshold": 0.30,
}
p30 = (proba >= 0.30).astype(int)
summary["recall_at_0.30"] = round(recall_score(y_test, p30), 3)
summary["precision_at_0.30"] = round(precision_score(y_test, p30), 3)
summary["top_drivers"] = imp.head(3).index.tolist()
for k, v in summary.items():
    print(f"{k:22s}: {v}")
```

### Recommendation

**Recommendation.** We can predict churn with a ROC-AUC of about 0.95 using only account and plan data that is known today (measured once on 600 held-out customers). At an alert threshold of 0.30 we catch about 84% of the customers who will leave, but only 73% of the customers we contact are real churners, so roughly one call in four goes to someone who would have stayed anyway; that is a good trade only while a saved customer is worth clearly more than a retention offer. The strongest drivers are cumulative spend (`total_charges`, which mostly reflects how long someone has been a customer), contract type and the number of support tickets, and the riskiest customers are on month-to-month contracts, pay by electronic check, use fibre and have many tickets. We recommend targeting month-to-month customers in their first year with proactive support and an incentive to move to an annual contract, and re-checking the 0.30 threshold on fresh data before it goes live (see the honesty note below).

### Reflection

**1. Leakage.** Another feature that could leak the outcome is something like `cancellation_reason`, `retention_offer_accepted` or `num_retention_calls`: it is only recorded *because* the customer already asked to cancel. To detect it before trusting a high score I would (a) ask for each column "is this value known at the moment we make the prediction?", (b) check how well each feature predicts the label on its own (here `exit_survey_sent` had a 0.95 correlation, far above everything else), and (c) look for one feature dominating the importance chart. A score that collapses when I train on older data and test on newer data is the final confirmation.

**2. Metric choice.** Precision matters more than recall when each alert is expensive or intrusive. For example, if the retention offer is a large discount (say 20% off for a year) or a personal call from an account manager with limited capacity (only 200 calls a week), then every false alarm gives away margin or wastes a scarce call, so I would raise the threshold and judge the model on precision among the top-ranked customers.

**3. Trust.** The four models scored 0.936 to 0.941, which is smaller than the fold-to-fold spread (the standard deviation was 0.012 to 0.019), so there is no real winner. I would deploy the simplest one that is easy to explain, the logistic regression, because its coefficients give signed, plain-language drivers and it trains instantly; the tuned gradient boosting model is not measurably better. It also tells me the ceiling comes from the data, not the algorithm: the next gain should come from better features (usage trends over time, billing events, the text of support tickets) or more data, not from a fancier model.

**4. Honesty.** If I choose the threshold on the test set and then report that same number, the test set is no longer unseen: I would be picking the threshold that happens to look best on these 600 customers, noise included, so the reported recall and precision would be optimistic and would not hold on new customers. I should own up that this notebook does a mild version of that: the 0.30 threshold was read from the sweep on the test set in Step 6. The cleaner way is to choose the threshold on the validation set (or with cross-validation), freeze it, and only then report the test result once.
